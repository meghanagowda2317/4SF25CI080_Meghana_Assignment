```python
import pandas as pd
import numpy as np
```

```python
df = pd.DataFrame(np.random.randint(1, 10, (5, 3)),
                  columns=['A', 'B', 'C'])
df['Product'] = df['A'] * df['B']
print(df)
```

```python
df = pd.DataFrame(np.random.randint(1, 10, (4, 3)),
                  columns=['A', 'B', 'C'])
print(df)
print("Row sum:")
print(df.sum(axis=1))
print("Column sum:")
print(df.sum(axis=0))
```

```python
df = pd.DataFrame(np.random.randint(1, 10, (5, 3)),
                  columns=['A', 'B', 'C'])
df.iloc[1, 0] = np.nan
df.iloc[3, 1] = np.nan
print(df)
df = df.fillna(df.mean())
print(df)
```

```python
df = pd.DataFrame(np.random.randint(1, 10, (6, 4)),
                  columns=['A', 'B', 'C', 'D'])
df.iloc[1, 2] = np.nan
df.iloc[3, 1] = np.nan
print(df)
df = df.dropna()
print(df)
```

```python
df = pd.DataFrame({
    'Category': np.random.choice(['A', 'B', 'C'], 10),
    'Value': np.random.randint(1, 100, 10)
})
print(df)
print(df.groupby('Category')['Value'].agg(['sum', 'mean']))
```

```python
df = pd.DataFrame({
    'Product': ['Pen', 'Book', 'Pencil', 'Bag', 'Box'],
    'Category': ['A', 'B', 'A', 'B', 'A'],
    'Sales': np.random.randint(100, 1000, 5)
})
print(df)
print(df.groupby('Category')['Sales'].sum())
```

```python
df1 = pd.DataFrame({
    'ID': [1, 2, 3],
    'Name': ['A', 'B', 'C']
})

df2 = pd.DataFrame({
    'ID': [1, 2, 3],
    'Marks': [80, 90, 85]
})

result = pd.merge(df1, df2, on='ID')
print(result)
```

```python
index = pd.MultiIndex.from_tuples([
    ('A', 1), ('A', 2), ('B', 1), ('B', 2)
])

df = pd.DataFrame({'Value': [10, 20, 30, 40]}, index=index)

print(df)
print(df.loc['A'])
print(df.loc['A':'B'])
```

```python
df = pd.DataFrame({
    'Category': ['A', 'A', 'B', 'B', 'C', 'C'],
    'SubCategory': ['X', 'Y', 'X', 'Y', 'X', 'Y'],
    'Value': np.random.randint(1, 100, 6)
})

df = df.set_index(['Category', 'SubCategory'])
print(df)
print(df.groupby(level=['Category', 'SubCategory']).sum())
```

```python
df = pd.DataFrame(np.random.randint(1, 10, (5, 3)),
                  columns=['A', 'B', 'C'])

result = df.apply(lambda x: x * 2)
print(result)
```

```python
df = pd.DataFrame(np.random.randint(1, 10, (6, 3)),
                  columns=['A', 'B', 'C'])

df['Sum'] = df.apply(lambda x: x['A'] + x['B'] + x['C'], axis=1)
print(df)
```
